```python
#Imports

import pandas as pd
import os
from datetime import datetime
import glob

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

```python
# Paths
RAW_PATH = "../data/raw/"
BRONZE_PATH = "../data/bronze/"


print(f"Raw path: {RAW_PATH}")
print(f"Bronze path: {BRONZE_PATH}")

# Verifying that raw files exist
print("\nRaw files found:")
for f in os.listdir(RAW_PATH):
    if f.endswith('.xlsx'):
        print(f"  - {f}")
```

```python
#Loading certifications
print("="*60)
print("BRONZE: Certifications")
print("="*60)

df = pd.read_excel(os.path.join(RAW_PATH, "Thomas More Data Certifications.xlsx"))
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")

# Add metadata
df['bronze_timestamp'] = datetime.now()
df['bronze_source'] = "Thomas More Data Certifications.xlsx"

# Save
output_path = os.path.join(BRONZE_PATH, "certifications_bronze.csv")
df.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
print(f"Shape saved: {df.shape}")
```

```python
#Loading clubs
print("="*60)
print("BRONZE: Clubs")
print("="*60)

df = pd.read_excel(os.path.join(RAW_PATH, "Thomas More Data Clubs.xlsx"))
print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")

# Add metadata
df['bronze_timestamp'] = datetime.now()
df['bronze_source'] = "Thomas More Data Clubs.xlsx"

# Save
output_path = os.path.join(BRONZE_PATH, "clubs_bronze.csv")
df.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
```

```python
#Loading results
print("="*60)
print("BRONZE: Results (all years)")
print("="*60)

# Find all result files
result_files = glob.glob(os.path.join(RAW_PATH, "Thomas More Results *.xlsx"))
print(f"Found {len(result_files)} files")

all_dfs = []
for file_path in result_files:
    filename = os.path.basename(file_path)
    year = filename.replace("Thomas More Results ", "").replace(".xlsx", "")
    
    df = pd.read_excel(file_path)
    df['source_year'] = year
    all_dfs.append(df)
    print(f"  {year}: {df.shape}")

# Combine
combined = pd.concat(all_dfs, ignore_index=True)
combined['bronze_timestamp'] = datetime.now()
combined['bronze_source'] = "Multiple Results files"

print(f"\nCombined shape: {combined.shape}")
print(f"Years: {sorted(combined['source_year'].unique())}")

# Save
output_path = os.path.join(BRONZE_PATH, "results_bronze.csv")
combined.to_csv(output_path, index=False)
print(f"Saved to: {output_path}")
```

```python
#validation check
print("="*60)
print("BRONZE VALIDATION")
print("="*60)

for f in os.listdir(BRONZE_PATH):
    if f.endswith('.csv'):
        df = pd.read_csv(os.path.join(BRONZE_PATH, f))
        print(f"{f}: {len(df)} rows, {len(df.columns)} columns")
```
